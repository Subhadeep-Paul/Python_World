```python
import numpy as np
import time
```

Using Python list to iterate over each number and calculate the sum and average.

```python
numbers = [20 , 22, 24, 26, 28, 30]
sum = 0;
for i in numbers:
  sum = sum + i
print("The total sum is:" , sum)
print("The average of numbers is:" , int(sum/len(numbers)))
```

Now we will be taking help of the NumPy library, Calculation is fast and memory efficient.

```python
sum = np.array(numbers).sum()
print("The total sum is:" , sum)
print("The average of numbers is:" , int(np.average(numbers)))
```

###Creating arrays in NumPy from scratch.
Zeros & Ones functions available in NumPy.

```python
array_zero = np.zeros((2,3))
print ("Zeros array: \n" , array_zero)


array_ones = np.ones((3,2))
print ("Ones array: \n" ,array_ones)


array_random = np.random.random((3,3))
print ("Random array: \n" ,array_random)


array_full = np.full((5,5), 4)
print ("Full array: \n" ,array_full)

#Or you can create it by multiplying a specific value with the ones array
#But notice the difference of output with the similar full function array
array_full4 = 4 * np.ones((5,5))
print ("Full array: \n" ,array_full4)


array_sequence = np.arange(1, 50, 3)
print ("Sequence array: \n" , array_sequence)


array_identity = np.eye(5) #Shape = 5 means 5by5 matrix
print ("Identity array: \n" , array_identity)
```

Now we will look into multiplication of NumPy array vs Python List. Closely watch the observation.

```python
print ("Multiplication of python list: " , numbers * 2)
print ("Multiplication of NumPy array: " , np.array(numbers))
```

In the multiplication of python list, the array itself got doubled because of the multiplication factor of 2. BUT, incase of NumPy Array, the values present in the array got multiplicated. Now, we'll look into the time required to calculate within the list/numpy array.

```python
start = time.time()
py_list = [i*2 for i in range(1,100000000)]
print ("Time taken by python list: " , (time.time() - start))

start = time.time()
numpy_array = np.arange(1,100000000) * 2
print ("Time taken by NumPy array: " , time.time() - start)
```

Notice the difference between the Python list and NumPy array. Python List have taken close to 11sec while the NumPy arrray have solved the same operation under 0.2 sec, which implies it is 50times faster than Python List.
